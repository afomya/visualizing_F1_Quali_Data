```python
import os
import fastf1
from fastf1 import plotting
import matplotlib.pyplot as plt

# Make sure the cache folder exists in Assignment_4
os.makedirs('f1_cache', exist_ok=True)
fastf1.Cache.enable_cache('f1_cache')

# Load session (example: Monza 2023 qualifying)
session = fastf1.get_session(2023, 'Monza', 'Q')
session.load()

# Get fastest lap in that session
lap = session.laps.pick_fastest()

# Extract telemetry
tel = lap.get_car_data().add_distance()
pos = lap.get_pos_data()

# Plot speed along the track
plt.scatter(pos['X'], pos['Y'], c=tel['Speed'], cmap='plasma', s=5)
plt.axis('equal')
plt.title(f"{lap['Driver']} – {session.event['EventName']} 2023 Qualifying")
plt.colorbar(label='Speed (km/h)')
plt.show()
```
